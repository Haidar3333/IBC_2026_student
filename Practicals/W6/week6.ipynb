```python
import pickle

genetic_code = pickle.load(open("../IntroBiolComp-2026/Python/genetic_code.pickle", "rb"))

test_mRNA = "AUGGAAUUCUCGCUCUGAAGGUAA"
```

```python
def get_amino_acids(mRNA):
    i = 0
    aa_sequence = []

    while (i + 3) < len(mRNA):
        codon = mRNA[i:(i + 3)]
        aa = genetic_code[codon]

        if aa == "Stop":
            break
        else:
            aa_sequence.append(aa)

        i = i + 4

    return "".join(aa_sequence)

print(get_amino_acids(test_mRNA))
```

```python
f = open(fasta_file, "r")
lines = f.readlines()
f.close()

print(len(lines))
print(lines[0])
print(lines[1])
```

```python
sequences = {}
header = ""

for line in lines:
    line = line.strip()

    if line.startswith(">"):
        header = line
        sequences[header] = ""
    else:
        sequences[header] += line

print(len(sequences))
```

```python
proteins = {}

for header, dna in sequences.items():
    mrna = dna.replace("T", "U")
    protein = get_amino_acids(mrna)
    proteins[header] = protein

print(len(proteins))
```

```python
output = open("Practicals/W6/Turkey_proteins.fasta", "w")

for header, protein in proteins.items():
    new_header = header.replace("gbkey=CDS", "gbkey=AA")
    output.write(new_header + "\n")
    output.write(protein + "\n")

output.close()
```
